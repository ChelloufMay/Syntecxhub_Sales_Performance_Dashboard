## 10. Transactional Grain Analysis

The dataset contains multiple rows for some orders.

This section determines the number of transaction lines associated with each Order ID and helps identify the appropriate grain for the future fact table.

```python
# Number of transaction lines per order

order_line_counts = (
    df.groupby("Order ID")
      .size()
      .sort_values(ascending=False)
)

print(f"Total transaction rows: {len(df):,}")
print(f"Unique orders: {df['Order ID'].nunique():,}")
print(f"Average transaction lines per order: {order_line_counts.mean():.2f}")
print(f"Minimum transaction lines for an order: {order_line_counts.min():,}")
print(f"Maximum transaction lines for an order: {order_line_counts.max():,}")
```

***The number of Unique orders (5,111) indicates that the grain cannot be one row per order.***

***Creating a fact table by assuming each row was a complete order would incorrectly treat one order containing five products as five separate orders, which would subsequently produce an incorrect KPI.***

```python
# Distribution of transaction lines per order

transaction_line_distribution = (
    order_line_counts
    .value_counts()
    .sort_index()
    .rename_axis("Transaction Lines per Order")
    .reset_index(name="Number of Orders")
)

transaction_line_distribution
```

```python
# Orders containing more than one transaction line

multi_line_orders = order_line_counts[order_line_counts > 1]

print(f"Orders with multiple transaction lines: {len(multi_line_orders):,}")
print(
    f"Percentage of orders with multiple lines: "
    f"{len(multi_line_orders) / len(order_line_counts) * 100:.2f}%"
)
```

```python
# Orders with the largest number of transaction lines

largest_orders = (
    order_line_counts
    .head(10)
    .reset_index(name="Transaction Lines")
)

largest_orders
```

The dataset contains 10,194 transaction rows and 5,111 unique orders.

Because the number of transaction rows is greater than the number of orders, and approximately 49.27% of orders contain multiple transaction lines, the dataset operates at a transaction/order-line grain rather than an order grain.

Each source row represents an individual transaction line associated with an Order ID. An order can therefore contain one or more transaction lines, with the largest observed order containing 14 lines.

Row ID contains 10,194 unique values and is therefore a strong candidate for uniquely identifying each transaction line.

This indicates that the transaction-line grain must be preserved during SQL transformation. The dataset shouldn't be aggregated to one row per Order ID before building the fact table because this would remove product-level, quantity, discount, sales, and profit information.

## 11. Identifier Relationship Analysis

Determine whether important identifiers consistently map to the same descriptive attributes.

The following relationships will be investigated:

- Customer ID → Customer Name
- Customer ID → Segment
- Product ID → Product Name

- Product ID → Sub-Category
- Order ID → Customer ID

These checks identify possible inconsistencies before SQL modeling.

### 11. 1. Customer Relationship Analysis

```python
# Customer ID consistency

customer_id_consistency = (
    df.groupby("Customer ID")
      .agg(
          Customer_Names=("Customer Name", "nunique"),
          Segments=("Segment", "nunique")
      )
      .sort_values(
          ["Customer_Names", "Segments"],
          ascending=False
      )
)

customer_id_consistency.head(20)
```

Each Customer ID consistently corresponds to:

one customer name
one segment

Conclusion: The customer information is consistent. We can safely use Customer ID as the customer identifier when creating the customer dimension.

```python
# Customer IDs with multiple names or segments

inconsistent_customers = customer_id_consistency[
    (customer_id_consistency["Customer_Names"] > 1) |
    (customer_id_consistency["Segments"] > 1)
]

print(
    f"Customer IDs with inconsistent name/segment mappings: "
    f"{len(inconsistent_customers):,}"
)

inconsistent_customers
```

```python
# Customer names associated with multiple Customer IDs

customer_name_consistency = (
    df.groupby("Customer Name")["Customer ID"]
      .nunique()
      .sort_values(ascending=False)
)

multiple_ids_same_name = customer_name_consistency[
    customer_name_consistency > 1
]

print(
    f"Customer names associated with multiple Customer IDs: "
    f"{len(multiple_ids_same_name):,}"
)

multiple_ids_same_name.head(20)
```

The name Harry Olson appears with 5 different Customer IDs.

This does not necessarily mean the data is wrong. It could represent different customers who happen to have the same name, or that he has multiple acoounts.

Conclusion: We should not merge customers based on their names. Customer ID is the reliable identifier, not Customer Name.

***Overall conclusion***

Customer ID → 1 Name + 1 Segment
Customer Name → can correspond to multiple Customer IDs

Therefore, for our SQL model: **Use Customer ID as the primary business identifier for customers.**

### 11. 2. Product Relationship Analysis

```python
# Product ID consistency

product_id_consistency = (
    df.groupby("Product ID")
      .agg(
          Product_Names=("Product Name", "nunique"),
          Categories=("Category", "nunique"),
          Sub_Categories=("Sub-Category", "nunique")
      )
      .sort_values(
          ["Product_Names", "Categories", "Sub_Categories"],
          ascending=False
      )
)

product_id_consistency.head(20)
```

```python
# Product IDs with inconsistent descriptive attributes

inconsistent_products = product_id_consistency[
    (product_id_consistency["Product_Names"] > 1) |
    (product_id_consistency["Categories"] > 1) |
    (product_id_consistency["Sub_Categories"] > 1)
]

print(
    f"Product IDs with inconsistent product information: "
    f"{len(inconsistent_products):,}"
)

inconsistent_products
```

***The results show 32 Product IDs linked to 2 different Product Names. This indicates that the same Product ID has been recorded with different names, while its category and sub-category stay the same.***

```python
# Product names associated with multiple Product IDs

product_name_consistency = (
    df.groupby("Product Name")["Product ID"]
      .nunique()
      .sort_values(ascending=False)
)

multiple_ids_same_product = product_name_consistency[
    product_name_consistency > 1
]

print(
    f"Product names associated with multiple Product IDs: "
    f"{len(multiple_ids_same_product):,}"
)

multiple_ids_same_product.head(20)
```

The results here show that there are 17 product names associated with multiple Product IDs.

Meaning, the relationship is not simply: Product ID ↔ Product Name

--> There are some inconsistencies in the product identifiers/names.

This indicates that we need to investigate the 32 inconsistent Product IDs and determine which product name should be retained when creating dim_product later on.

Because `Category` and `Sub-Category` are consistent for those 32 IDs, the problem appears to be specifically with the Product Name, not the product classification.

**Conclusion:**

Product ID is not perfectly consistent with Product Name in the raw dataset. Therefore, product relationships must be validated and standardized during the SQL cleaning stage before creating the product dimension. The rows should not be removed.

## 13. Numerical Data Quality Checks

The following checks identify values that may require investigation.

```python
quality_checks = {}

if "Sales" in df.columns:
    quality_checks["Negative Sales"] = int(
        (df["Sales"] < 0).sum()
    )

if "Quantity" in df.columns:
    quality_checks["Non-positive Quantity"] = int(
        (df["Quantity"] <= 0).sum()
    )

if "Discount" in df.columns:
    quality_checks["Discount < 0"] = int(
        (df["Discount"] < 0).sum()
    )
    
    quality_checks["Discount > 1"] = int(
        (df["Discount"] > 1).sum()
    )

if "Profit" in df.columns:
    quality_checks["Negative Profit"] = int(
        (df["Profit"] < 0).sum()
    )

numerical_quality_summary = pd.DataFrame(
    list(quality_checks.items()),
    columns=["Check", "Number of Rows"]
)

numerical_quality_summary
```

`Row ID`: is unique for every row and is a good candidate for the unique identifier of a transaction line.

`Sales`: The mean is much higher than the median, showing that sales are strongly right-skewed: most transactions are relatively small, while a smaller number have very high sales.

`Quantity`: All transactions have a positive quantity. There are no obvious invalid quantities.

`Discount`: Discount values range from 0% to 80%. The check found no values below 0 or above 1, so there are no obviously invalid discount values.

`Profit`: Profit varies significantly. 1,901 rows have negative profit, which is about 18.65% of all transaction rows. That is not automaticlly a data-quality. It means those transactions generated losses.

***Overall numerical conclusion***

The numerical data appears structurally valid. No negative sales, zero/negative quantities, or invalid discount values were found. Profit contains legitimate negative values that should be retained because they represent loss-making transactions.

## 1. Dataset Structure

The raw dataset contains 10,194 transaction rows and 21 columns.

It includes information about orders, customers, products, shipping, geography and financial performance.

The dataset is suitable for building a sales performance analytics solution.

## 2. Transactional Grain

There are 5,111 unique Order IDs across 10,194 transaction rows.

The average order contains approximately 1.99 transaction lines, and 2,518 orders (49.27%) contain more than one transaction line.

The largest observed order contains 14 transaction lines.

Row ID contains 10,194 unique values, matching the number of transaction rows. Therefore, Row ID is a strong candidate for uniquely identifying each transaction line.

The fact table must preserve the transaction-line grain rather than aggregating the raw data to one row per Order ID.

---

## 3. Missing Values

No missing values were found in any column.

No rows need to be removed or values filled because of missing data.

SQL validation will still confirm this during the staging stage.

---

## 4. Duplicate Records

No exact duplicate rows were found.

Repeated Order IDs are not considered duplicate rows because an order can contain multiple transaction lines for different products.

No duplicate-row removal is required.

---

## 5. Data Types

The following data-type changes will be required during SQL transformation:

- Order Date: convert from text to DATE.
- Ship Date: convert from text to DATE.
- Postal Code: keep as text because it is an identifier and can contain letters.

The other fields will be assigned appropriate SQL data types during staging.

---

## 6. Customer Data

There are 804 unique Customer IDs and 800 unique Customer Names.

Each Customer ID consistently maps to one Customer Name and one Segment.

However, one customer name, Harry Olson, appears with five different Customer IDs.

Therefore, Customer Name must not be used as the customer key.

Customer ID will be used as the reliable customer business identifier.

No customer rows should be removed because of the repeated name.

---

## 7. Product Data

There are 1,862 unique Product IDs and 1,849 unique Product Names.

32 Product IDs are associated with two different Product Names.

Their Category and Sub-Category remain consistent, so the inconsistency is specifically in Product Name.

There are also 17 Product Names associated with multiple Product IDs.

Therefore, Product ID and Product Name cannot be treated as a perfect one-to-one relationship in the raw data.

The 32 inconsistent Product IDs must be investigated and standardized during SQL cleaning before creating the product dimension.

The affected rows should not be deleted.

Category and Sub-Category should be validated and preserved.

---

## 8. Order Consistency

The Order ID correctly groups multiple transaction lines.

Order Date is consistent within each order, with zero orders associated with multiple Order Dates.

Ship Date is also consistent within each order, with zero orders associated with multiple Ship Dates.

However, 2 orders are associated with multiple Customer IDs.

These 2 orders must be investigated during SQL data-quality validation before the final dimensional model is created.

No rows should be removed at the profiling stage.

---

## 9. Numerical Data

Row ID is unique for every transaction line.

Sales contains no negative values.

Quantity contains no zero or negative values.

Discount contains no values below 0 or above 1.

Profit contains 1,901 negative-profit rows, representing approximately 18.65% of transaction rows.

Negative Profit is not treated as a data-quality error because it represents loss-making transactions and is important for profitability analysis.

These rows must therefore be retained.

Ship Date ranges from 2023-01-07 to 2027-01-05.

All dates were successfully interpreted during profiling.

No shipment occurs before its corresponding order.

A derived shipping-duration measure can later be used for shipping-performance analysis.
